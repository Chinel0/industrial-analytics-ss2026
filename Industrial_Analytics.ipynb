### Data Organization into Dictionary 

```python
import pandas as pd
import numpy as np

# Loading datasets
train_set = pd.read_csv(r"C:\Users\nweke\Downloads\train_set.csv", index_col=0)
bill_of_materials = pd.read_csv(r"C:\Users\nweke\Downloads\bill_of_materials.csv", index_col=0)
tube = pd.read_csv(r"C:\Users\nweke\Downloads\tube.csv", index_col=0)
specs = pd.read_csv(r"C:\Users\nweke\Downloads\specs.csv", index_col=0)

datasets = {
    "Train Set": train_set,
    "BOM": bill_of_materials,
    "Tube": tube,
    "Specs": specs
}
```

### Conformity Check

```python
for name, df in datasets.items():
    print(f"--- {name} Quality Check ---")
    print(f"Data Types:\n{df.dtypes}\n")
```

### Completeness: Check for NULL values

```python
#  Completeness: Count and show percentage of missing values
for name, df in datasets.items():
    missing = df.isnull().mean() * 100
    print(f"--- {name} Missing Data (%) ---")
    print(missing[missing > 0] if missing.any() else "No missing values")
    print("-" * 20)
```

### Duplicate data (Uniqueness Issue)

```python
for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name} has {duplicates} duplicate rows.")
    
    # Optional: View duplicates if they exist
    if duplicates > 0:
        print(df[df.duplicated(keep=False)].head())
```

### Inaccurate data (Accuracy issue)

```python
def check_outliers(df, name):
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        if not outliers.empty:
            print(f"Potential outliers in {name}['{col}']: {len(outliers)} records found.")

check_outliers(tube, "Tube")
check_outliers(train_set, "Train Set")
```

### Out-of-date data (Timeliness Issue)

```python
# Timeliness: Check date range in train_set (if 'quote_date' exists)
if 'quote_date' in train_set.columns:
    train_set['quote_date'] = pd.to_datetime(train_set['quote_date'])
    print(f"Data Date Range: {train_set['quote_date'].min()} to {train_set['quote_date'].max()}")
```

### Invalid data (Consistency Issue)

```python
# (6) Invalid data / Consistency issue
# Detect inconsistent text values such as extra spaces, mixed casing, and label variants

for name, df in datasets.items():
    print(f"\n{'='*10} Dataset: {name} {'='*10}")
    
    # Identify text-based columns
    text_cols = df.select_dtypes(include=["object", "string", "category"]).columns.tolist()
    
    if not text_cols:
        print("No text columns found.")
        continue
    
    for col in text_cols:
        s = df[col].dropna().astype(str)
        if s.empty: continue
        
        # 1. Check for invisible spaces
        leading_trailing_spaces = (s != s.str.strip()).sum()
        
        # 2. Check for inconsistent naming (e.g., 'Steel' and 'steel' in same column)
        # We group by the 'lowercase' version and count how many 'original' versions exist
        normalized = s.str.strip().str.lower()
        variant_counts = (
            pd.DataFrame({"original": s, "normalized": normalized})
            .groupby("normalized")["original"]
            .nunique()
        )
        inconsistent_labels = variant_counts[variant_counts > 1]
        
        print(f"Column: [{col}]")
        print(f"  - Rows with extra spaces: {leading_trailing_spaces}")
        print(f"  - Normalized duplicates: {normalized.duplicated().sum()}")
        
        if not inconsistent_labels.empty:
            print(f"  - ALERT: Found {len(inconsistent_labels)} labels with inconsistent casing/spacing!")
            print("    Top offenders:")
            print(inconsistent_labels.head(3))
        print("-" * 30)
```

### Supplier Pricing Variability Analysis

```python
# 3.7 Supplier Pricing Variability Analysis

# Supplier Price Variability for Same Tube Characteristics
print("=" * 60)
print("SUPPLIER PRICING VARIABILITY ANALYSIS")
print("=" * 60)

# Since index_col=0 was used, we group by the index, not a column name
# We check how many unique suppliers appear for each index (tube_assembly_id)
tube_supplier_counts = train_merged.groupby(level=0)['supplier'].nunique()
multi_supplier_tubes = tube_supplier_counts[tube_supplier_counts > 1].index

if len(multi_supplier_tubes) > 0:
    # Use .loc to find rows where the index is in our multi-supplier list
    multi_quoted = train_merged.loc[multi_supplier_tubes]
    print(f"\nTubes quoted by multiple suppliers: {len(multi_supplier_tubes)}")
    print(f"Total quotes for these tubes: {len(multi_quoted)}")
    
    # Price variation stats (grouping by level 0 which is the index)
    price_variation = multi_quoted.groupby(level=0)['cost'].agg(['min', 'max', 'mean', 'std'])
    price_variation['variation_pct'] = ((price_variation['max'] - price_variation['min']) / price_variation['mean'] * 100)
    
    print(f"\nPrice Variation Statistics (% difference between min and max supplier):")
    print(f"Average variation: {price_variation['variation_pct'].mean():.2f}%")
    print(f"Max variation: {price_variation['variation_pct'].max():.2f}%")
    print(f"Min variation: {price_variation['variation_pct'].min():.2f}%")
    
    # Visualization
    fig, ax = plt.subplots(figsize=(12, 6))
    price_variation['variation_pct'].hist(bins=30, ax=ax, edgecolor='black', color='lightblue')
    ax.set_title('Distribution of Price Variation Across Suppliers\n(for tubes with multiple quotes)', fontsize=12, fontweight='bold')
    ax.set_xlabel('Price Variation (%)')
    ax.set_ylabel('Frequency')
    
    mean_val = price_variation['variation_pct'].mean()
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}%')
    ax.legend()
    plt.tight_layout()
    plt.show()
else:
    print("No tubes found with multiple supplier quotes in this dataset.")
```

## Determinig Total quanitiy and weight of tube from Dataset bill of material

There are a total of 11 comp_[name] files, which contains the details for each component including weight.

In general, the heavier a tube is, the more expensive it can be. In addition, the more quantity required, the more expensive a tube can be.

Thus, we'll try to extract the total quantity and total weight features for each tube.

```python
bill_of_materials.head()
```

```python
qty_cols = [c for c in bill_of_materials.columns if c.startswith("quantity_")]
id_cols = [c for c in bill_of_materials.columns if c.startswith("component_id_")]

bill_of_materials[qty_cols] = bill_of_materials[qty_cols].fillna(0)
bill_of_materials[id_cols] = bill_of_materials[id_cols].fillna("NONE")
```

```python
# Starting from 'quantity_1', we sum up quantities every other columns
bill_of_materials['Total_quantity'] = bill_of_materials.iloc[:, 2::2].sum(axis=1) 
```

```python
bill_of_materials.head()
```

Now, let's move onto weight feature

```python
# we use for in loop to select comp name from the bracket and import its dataset in data from comp_df and than only pick the data related to comp_id and its weight in a new dataframe called camp_dfs
#This function loops through multiple component CSV files, extracts component IDs and weights, and merges them into one unified DataFrame.

def get_comp_weight():
    # 1. Names based on your folder (image_aaf6d2.png)
    comp_names = ['adaptor', 'boss', 'elbow', 'float', 'hfl', 'nut', 
                  'other', 'sleeve', 'straight', 'tee', 'threaded']

    comp_dfs = pd.DataFrame()
    for comp_name in comp_names:
        # 2. Path updated to your specific folder
        comp_df = pd.read_csv(r"C:\Users\nweke\Downloads\comp_" + comp_name + ".csv")
        comp_dfs = pd.concat([comp_dfs, comp_df[['component_id', 'weight']]])

    return comp_dfs

# To run it and see the result:
all_comp_weights = get_comp_weight()
all_comp_weights.head()
```

```python
comp_dfs = get_comp_weight()
```

```python
comp_dfs.info()
```

```python
# Fill missing values with 0 in weigt columns
comp_dfs['weight'] = comp_dfs['weight'].fillna(0)
```

```python
# we used a look to Replace component_id columns with respective weight
for i in range(1,9):
    comp_no = 'component_id_'+str(i)
    bill_of_materials[comp_no] = bill_of_materials[comp_no].map(comp_dfs.set_index('component_id')['weight'])
```

```python
bill_of_materials.sample(10, random_state=42)
```

```python
bill_of_materials.fillna(0, inplace=True)

# Calculate total weight for each tube

bill_of_materials['Total_weight'] = 0
for i in range(1,9):
    comp_no = 'component_id_'+ str(i)
    quant_no = 'quantity_'+ str(i)
    
    bill_of_materials['Total_weight'] = bill_of_materials['Total_weight'] + bill_of_materials[comp_no]*bill_of_materials[quant_no]
```
