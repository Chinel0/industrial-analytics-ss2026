## Determinig Total quanitiy and weight of tube from Dataset bill of material

There are a total of 11 comp_[name] files, which contains the details for each component including weight.

In general, the heavier a tube is, the more expensive it can be. In addition, the more quantity required, the more expensive a tube can be.

Thus, we'll try to extract the total quantity and total weight features for each tube.

```python
bom.head()
```

```python
bom = bom.fillna(0)
```

```python
# Starting from 'quantity_1', we sum up quantities every other columns
bom['Total_quantity'] = bom.iloc[:, 2::2].sum(axis=1) 
```

### 
The code uses bom.iloc[:, 2::2] to select every other column starting from column index 2 (which captures all the quantity_1, quantity_2, quantity_3... columns), then .sum(axis=1) adds them up across each row to get the total quantity of components in each tube assembly.

```python
bom.head()
```

Now, let's move onto weight feature

```python
# we use for in loop to select comp name from the bracket and import its dataset in data from comp_df and than only pick the data related to comp_id and its weight in a new dataframe called camp_dfs
#This function loops through multiple component CSV files, extracts component IDs and weights, and merges them into one unified DataFrame.

def get_comp_weight():
    
    comp_names = ['adaptor', 'boss', 'elbow', 'float', 
                  'hfl', 'nut', 'other', 'sleeve',
                  'straight', 'tee', 'threaded']

    comp_dfs = pd.DataFrame()
    for comp_name in comp_names:
         comp_df = pd.read_csv(r"C:\Users\nweke\Downloads\comp_" + comp_name + ".csv")
         comp_dfs = pd.concat([comp_dfs, comp_df[['component_id', 'weight']]])

    return comp_dfs
# To run it and see the result:
all_comp_weights = get_comp_weight()
all_comp_weights.head()
```

```python
comp_dfs = get_comp_weight()
```

```python
comp_dfs.info()
```

```python
# Fill missing values with 0 in weigt columns
comp_dfs['weight'] = comp_dfs['weight'].fillna(0)
```

```python
# we used a look to Replace component_id columns with respective weight
for i in range(1,9):
    comp_no = 'component_id_'+str(i)
    bom[comp_no] = bom[comp_no].map(comp_dfs.set_index('component_id')['weight'])
```

```python
bom.sample(10, random_state=42)
```

```python
bom.fillna(0, inplace=True)

# Calculate total weight for each tube

bom['Total_weight'] = 0
for i in range(1,9):
    comp_no = 'component_id_'+ str(i)
    quant_no = 'quantity_'+ str(i)
    
    bom['Total_weight'] = bom['Total_weight'] + bom[comp_no]*bom[quant_no]
```

###  First, bom.fillna(0, inplace=True) replaces all missing values in the bill-of-materials dataframe with 0 (since missing component weights/quantities represent "no component"). Then, the loop calculates the total weight of each tube assembly by multiplying each of the 8 component weights by their quantities and summing them together.

## Import Models

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, cross_validate, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline

import joblib
import warnings
warnings.filterwarnings('ignore')
```

###  Phase 1: Feature Preparation & 3-Way Data Split
train (60%) | validation (20%) | test (20%)
The TEST SET is sealed here and never touched until Phase 5.
All tuning and cross-validation happen on the training set only.

Drop redundant index columns and define X (features) and y (target).
The target variable is `cost` — the supplier price we want to predict.

```python
cols_to_drop = ['Unnamed: 0', 'index_x', 'index_y', 'year_group',
                'end_a', 'end_x', 'tube_prefix']
cols_to_drop = [c for c in cols_to_drop if c in Final_DF.columns]
df_model = Final_DF.drop(columns=cols_to_drop)

X = df_model.drop(columns=['cost'])
y = df_model['cost']

# Log-transform target — cost is heavily right-skewed
y_log = np.log1p(y)

print(f"Total samples : {X.shape[0]}")
print(f"Total features: {X.shape[1]}")
print(f"\nFeatures:\n{list(X.columns)}")

# Step 1 — hold out 20% as the truly unseen test set
X_trainval, X_test, y_trainval, y_test = train_test_split(
    X, y_log, test_size=0.2, random_state=42
)

# Step 2 — split remaining 80% into train (75% of 80% = 60%) and validation (20%)
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval, test_size=0.25, random_state=42
)

print(f"\n{'─'*45}")
print(f"  Training set   : {X_train.shape[0]:>5} rows  ({X_train.shape[0]/len(X)*100:.0f}%)")
print(f"  Validation set : {X_val.shape[0]:>5} rows  ({X_val.shape[0]/len(X)*100:.0f}%)")
print(f"  Test set       : {X_test.shape[0]:>5} rows  ({X_test.shape[0]/len(X)*100:.0f}%)  ← sealed")
print(f"{'─'*45}")
```

### Helper: Evaluate on Original Dollar Scale

```python
def evaluate(pipeline, X, y_log_true, label=""):
    """Predict and score, converting log-predictions back to original $ scale."""
    y_pred_log  = pipeline.predict(X)
    y_true_orig = np.expm1(y_log_true)
    y_pred_orig = np.expm1(y_pred_log)
    mae  = mean_absolute_error(y_true_orig, y_pred_orig)
    rmse = np.sqrt(mean_squared_error(y_true_orig, y_pred_orig))
    r2   = r2_score(y_true_orig, y_pred_orig)
    if label:
        print(f"  {label:45s}  MAE=${mae:7.4f}  RMSE=${rmse:7.4f}  R²={r2:.4f}")
    return {"MAE": mae, "RMSE": rmse, "R²": r2,
            "y_true": y_true_orig, "y_pred": y_pred_orig}
```

### Phase 2: Baseline Linear Regression, No regularisation. Evaluated on the validation set

### Model A: Linear Regression on Raw Cost

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline

# Raw (untransformed) targets for Model A
y_train_raw = np.expm1(y_train)
y_val_raw   = np.expm1(y_val)

model_A = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  LinearRegression())
])
model_A.fit(X_train, y_train_raw)

# Evaluate on validation set
y_pred_A = model_A.predict(X_val)

mae_A  = mean_absolute_error(y_val_raw, y_pred_A)
rmse_A = np.sqrt(mean_squared_error(y_val_raw, y_pred_A))
r2_A   = r2_score(y_val_raw, y_pred_A)

print("=== Model A — Linear Regression (Raw Cost) ===")
print(f"Validation MAE  : ${mae_A:.4f}")
print(f"Validation RMSE : ${rmse_A:.4f}")
print(f"Validation R²   : {r2_A:.4f}")
```

### Model A: 5-Fold Cross-Validation

```python
from sklearn.model_selection import KFold, cross_validate

kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_A = cross_validate(
    model_A, X_train, y_train_raw,
    cv=kf,
    scoring={
        'r2'  : 'r2',
        'mae' : 'neg_mean_absolute_error',
        'rmse': 'neg_root_mean_squared_error'
    }
)

cv_mae_A  = -cv_A['test_mae'].mean()
cv_rmse_A = -cv_A['test_rmse'].mean()
cv_r2_A   =  cv_A['test_r2'].mean()

print("=== Model A — 5-Fold Cross-Validation (Raw Cost) ===")
print(f"CV MAE  : ${cv_mae_A:.4f}")
print(f"CV RMSE : ${cv_rmse_A:.4f}")
print(f"CV R²   : {cv_r2_A:.4f}")
print(f"\nAll CV R² scores: {cv_A['test_r2'].round(3)}")
```

### Model B: Linear Regression on Log-Transformed Cost
log1p applied to cost to reduce right skew
y_train and y_val are already log-transformed from Phase 1


```python
model_B = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  LinearRegression())
])
model_B.fit(X_train, y_train)

# Predict and convert back to original dollar scale
y_pred_B_log = model_B.predict(X_val)
y_pred_B     = np.expm1(y_pred_B_log)
y_val_orig   = np.expm1(y_val)

mae_B  = mean_absolute_error(y_val_orig, y_pred_B)
rmse_B = np.sqrt(mean_squared_error(y_val_orig, y_pred_B))
r2_B   = r2_score(y_val_orig, y_pred_B)

print("=== Model B — Linear Regression (Log-Transformed Cost) ===")
print(f"Validation MAE  : ${mae_B:.4f}")
print(f"Validation RMSE : ${rmse_B:.4f}")
print(f"Validation R²   : {r2_B:.4f}")
```

### Model B: 5-Fold Cross-Validation
Metrics are on the log scale (as trained)

```python
cv_B = cross_validate(
    model_B, X_train, y_train,
    cv=kf,
    scoring={
        'r2'  : 'r2',
        'mae' : 'neg_mean_absolute_error',
        'rmse': 'neg_root_mean_squared_error'
    }
)

cv_mae_B  = -cv_B['test_mae'].mean()
cv_rmse_B = -cv_B['test_rmse'].mean()
cv_r2_B   =  cv_B['test_r2'].mean()

print("=== Model B — 5-Fold Cross-Validation (Log Scale) ===")
print(f"CV MAE  : {cv_mae_B:.4f}  (log scale)")
print(f"CV RMSE : {cv_rmse_B:.4f}  (log scale)")
print(f"CV R²   : {cv_r2_B:.4f}")
print(f"\nAll CV R² scores: {cv_B['test_r2'].round(3)}")
```

### Model B: Actual vs Predicted Chart

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_val_orig, y_pred_B, alpha=0.3, color='seagreen', edgecolors='none', s=15)
plt.plot([y_val_orig.min(), y_val_orig.max()],
         [y_val_orig.min(), y_val_orig.max()],
         'r--', linewidth=1.5, label='Perfect prediction')
plt.xlabel('Actual Cost ($)')
plt.ylabel('Predicted Cost ($)')
plt.title('Model B — Actual vs Predicted (Log-Transformed Cost)')
plt.legend()
plt.tight_layout()
plt.savefig('model_B_actual_vs_predicted.png', dpi=150)
plt.show()
```

### Phase 3: Hyperparameter Tuning with GridSearchCV
GridSearchCV tests every alpha value using K-fold CV on
 X_train only. The validation set is NOT seen here.
 We tune:
   • Ridge Regression (L2 regularisation)
   • Lasso Regression (L1 regularisation)


```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split, GridSearchCV, KFold, cross_validate, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline

print("="*60)
print("PHASE 3 — HYPERPARAMETER TUNING (GridSearchCV)")
print("="*60)

kf = KFold(n_splits=5, shuffle=True, random_state=42)

# ── Ridge ──────────────────────────────────────────────────
ridge_pipe   = Pipeline([('scaler', StandardScaler()), ('model', Ridge())])
ridge_params = {'model__alpha': [0.01, 0.1, 1, 10, 100, 1000]}

ridge_gs = GridSearchCV(
    ridge_pipe, ridge_params,
    cv=kf, scoring='neg_mean_absolute_error',
    return_train_score=True, n_jobs=-1
)
ridge_gs.fit(X_train, y_train)

# ── Lasso ──────────────────────────────────────────────────
lasso_pipe   = Pipeline([('scaler', StandardScaler()),
                          ('model', Lasso(max_iter=10000))])
lasso_params = {'model__alpha': [0.0001, 0.001, 0.01, 0.1, 1]}

lasso_gs = GridSearchCV(
    lasso_pipe, lasso_params,
    cv=kf, scoring='neg_mean_absolute_error',
    return_train_score=True, n_jobs=-1
)
lasso_gs.fit(X_train, y_train)

# ── Print tuning trial tables ───────────────────────────────
def tuning_table(gs, name):
    df = pd.DataFrame(gs.cv_results_)
    col = [c for c in df.columns if 'param_model__alpha' in c][0]
    out = df[[col, 'mean_test_score', 'std_test_score']].copy()
    out.columns = ['alpha', 'CV MAE (neg)', 'std']
    out['CV MAE (log scale)'] = -out['CV MAE (neg)']
    out = out[['alpha', 'CV MAE (log scale)', 'std']].sort_values('CV MAE (log scale)')
    print(f"\n{name} — all {len(out)} trials:")
    print(out.to_string(index=False))
    print(f"→ Best alpha = {gs.best_params_['model__alpha']}  "
          f"(CV MAE = {-gs.best_score_:.4f})")

tuning_table(ridge_gs, "Ridge Regression")
tuning_table(lasso_gs, "Lasso Regression")
```

### Tuning Visualisation

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, gs, color, label in [
    (axes[0], ridge_gs, 'steelblue', 'Ridge'),
    (axes[1], lasso_gs, 'seagreen',  'Lasso'),
]:
    alphas   = [p['model__alpha'] for p in gs.cv_results_['params']]
    cv_maes  = [-s for s in gs.cv_results_['mean_test_score']]
    cv_stds  = gs.cv_results_['std_test_score']
    best_a   = gs.best_params_['model__alpha']

    ax.semilogx(alphas, cv_maes, 'o-', color=color, linewidth=2, markersize=8)
    ax.fill_between(alphas,
                    [m - s for m, s in zip(cv_maes, cv_stds)],
                    [m + s for m, s in zip(cv_maes, cv_stds)],
                    alpha=0.15, color=color)
    ax.axvline(best_a, color='red', linestyle='--',
               label=f'Best α = {best_a}')
    ax.set_title(f'{label} Regression — Tuning Curve', fontsize=13, fontweight='bold')
    ax.set_xlabel('Alpha (regularisation strength, log scale)')
    ax.set_ylabel('CV MAE (log scale)')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.suptitle('Hyperparameter Tuning — 5-Fold Cross-Validation on Training Set',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('hyperparameter_tuning.png', dpi=150, bbox_inches='tight')
plt.show()
```

### PHASE 4 — VALIDATION ON HELD-OUT VALIDATION SET
We now apply the best parameters found in Phase 3 to the
validation set. This data was never used during tuning.
This confirms the chosen parameters actually generalise.


```python
res_base = {'MAE': mae_B, 'RMSE': rmse_B, 'R²': r2_B}
print("="*60)
print("PHASE 4 — VALIDATION ON HELD-OUT VALIDATION SET")
print("="*60)
print("(Validation set was sealed during all tuning in Phase 3)\n")

res_ridge = evaluate(ridge_gs.best_estimator_, X_val, y_val,
                     f"Ridge (best α={ridge_gs.best_params_['model__alpha']}) (validation)")
res_lasso = evaluate(lasso_gs.best_estimator_, X_val, y_val,
                     f"Lasso (best α={lasso_gs.best_params_['model__alpha']}) (validation)")

# Summary table — all metrics in original $ scale
val_summary = pd.DataFrame({
    'Model': [
        'Linear Regression (baseline)',
        f"Ridge — α={ridge_gs.best_params_['model__alpha']}",
        f"Lasso — α={lasso_gs.best_params_['model__alpha']}"
    ],
    'Val MAE ($)' : [res_base['MAE'],  res_ridge['MAE'],  res_lasso['MAE']],
    'Val RMSE ($)': [res_base['RMSE'], res_ridge['RMSE'], res_lasso['RMSE']],
    'Val R²'      : [res_base['R²'],   res_ridge['R²'],   res_lasso['R²']],
})

print("\nValidation Results (original $ scale — lower MAE/RMSE is better):")
print(val_summary.to_string(index=False))

# Select best model by lowest validation MAE
best_idx  = val_summary['Val MAE ($)'].idxmin()
best_name = val_summary.loc[best_idx, 'Model']
best_candidates = {
    0: ('linear',  None),
    1: ('ridge',   ridge_gs.best_params_['model__alpha']),
    2: ('lasso',   lasso_gs.best_params_['model__alpha']),
}
best_type, best_alpha = best_candidates[best_idx]
print(f"\n→ Best model selected: {best_name}")

# Bar chart comparing validation metrics
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
colors = ['steelblue', 'seagreen', 'darkorange']
models = val_summary['Model'].str.split(' —').str[0]

for i, (ax, metric) in enumerate(zip(axes, ['Val MAE ($)', 'Val RMSE ($)', 'Val R²'])):
    bars = ax.bar(models, val_summary[metric], color=colors, edgecolor='black', width=0.5)
    ax.bar_label(bars, fmt='%.3f', padding=3, fontsize=9)
    ax.set_title(metric, fontweight='bold')
    ax.set_ylabel(metric)
    ax.tick_params(axis='x', rotation=15)

plt.suptitle('Validation Set Comparison — All Models (Original $ Scale)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('validation_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

### PHASE 5 — FINAL EVALUATION ON TRULY UNSEEN TEST SET
 The test set has been sealed since Phase 1.
 We retrain the best model on train + validation combined,
 then evaluate ONCE on the test set.
 These are the only numbers we report as final results.


```python
print("="*60)
print("PHASE 5 — FINAL EVALUATION ON UNSEEN TEST SET")
print("="*60)

# Combine train + validation for final training
X_full_train = pd.concat([X_train, X_val])
y_full_train = pd.concat([y_train, y_val])

# Build final pipeline with best parameters
if best_type == 'linear':
    final_model = LinearRegression()
elif best_type == 'ridge':
    final_model = Ridge(alpha=best_alpha)
else:
    final_model = Lasso(alpha=best_alpha, max_iter=10000)

final_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  final_model)
])
final_pipeline.fit(X_full_train, y_full_train)

# ── Evaluate once on test set ───────────────────────────────
print(f"\nFinal model : {best_name}")
print(f"Trained on  : {len(X_full_train)} rows (train + validation)")
print(f"Tested on   : {len(X_test)} rows (previously unseen)\n")

res_final = evaluate(final_pipeline, X_test, y_test, f"FINAL — {best_name} (test set)")

# ── Stability check via CV on the full training data ────────
cv_r2 = cross_val_score(final_pipeline, X_full_train, y_full_train,
                        cv=kf, scoring='r2')
print(f"\n5-Fold CV R² on train+val (stability check): "
      f"{cv_r2.round(3)}  →  mean={cv_r2.mean():.4f} ± {cv_r2.std():.4f}")

# ── Actual vs Predicted plot ─────────────────────────────────
plt.figure(figsize=(8, 6))
plt.scatter(res_final['y_true'], res_final['y_pred'],
            alpha=0.3, color='steelblue', s=15, edgecolors='none')
min_val = min(res_final['y_true'].min(), res_final['y_pred'].min())
max_val = max(res_final['y_true'].max(), res_final['y_pred'].max())
plt.plot([min_val, max_val], [min_val, max_val],
         'r--', linewidth=1.5, label='Perfect prediction')
plt.xlabel('Actual Cost ($)', fontsize=12)
plt.ylabel('Predicted Cost ($)', fontsize=12)
plt.title(f'Final Model — Actual vs Predicted (Test Set)\n{best_name}', fontweight='bold')
plt.legend()
plt.tight_layout()
plt.savefig('final_model_actual_vs_predicted.png', dpi=150)
plt.show()

# ── Save model and predictions ───────────────────────────────
joblib.dump(final_pipeline, 'final_model.pkl')
pd.DataFrame({
    'actual_cost'   : res_final['y_true'].values,
    'predicted_cost': res_final['y_pred']
}).to_csv('final_predictions.csv', index=False)
print("\nSaved: final_model.pkl  |  final_predictions.csv")
```

### Feature Importance via Standardized Coefficients

### SHAP Global Feature Importance (Summary Plot)

```python
import shap

# LinearExplainer is the correct explainer for linear regression + StandardScaler pipeline
explainer = shap.LinearExplainer(
    model_B.named_steps['model'],
    shap.sample(model_B.named_steps['scaler'].transform(X_train), 100)
)

X_val_scaled = model_B.named_steps['scaler'].transform(X_val)
shap_values  = explainer.shap_values(X_val_scaled)

# Summary plot — shows which features have the most impact overall
shap.summary_plot(
    shap_values,
    X_val,
    feature_names=feature_names,
    plot_type='bar',
    show=False
)
plt.title('SHAP Feature Importance — Model B (Global)', fontweight='bold')
plt.tight_layout()
plt.savefig('shap_global_importance.png', dpi=150, bbox_inches='tight')
plt.show()
```

### SHAP Detailed Summary (shows direction too)

```python
# Dot plot — shows both importance AND direction of each feature's effect
shap.summary_plot(
    shap_values,
    X_val,
    feature_names=feature_names,
    show=False
)
plt.title('SHAP Summary — Feature Impact on Cost Prediction', fontweight='bold')
plt.tight_layout()
plt.savefig('shap_summary_dot.png', dpi=150, bbox_inches='tight')
plt.show()
```
